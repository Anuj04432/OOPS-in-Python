What is OOPs?
---
OOPS stands for Object Oriented Programming System

It is programming paradigm that organizes code using clases and objects.


Key concepts of OOPs
---
1.Class --> Its is a blueprint for crating an object

2.Object --> It's is an instance of an object

3.Constructor( __init__)--> It initializes the object's attributes

```python
# Example of class
class Car:
    pass

# Example of object
class Bank:
    pass

Sbi = Bank()
print(type(Sbi))


# Example of Constuctor
# class
class Atm:

    # Constructor
    def __init__(self):    # __init__ is also known as magic methods this are already pre defined
        self.pin = '1234'
        self.balance = 2000

# Object
atm = Atm()
atm.pin

```

Why we use self?
---
The __self__ refers to the current object(instance) of the class. It tells python which object's data or method you're reffering to.

Example without self
---

```python
class Student:
    def __init__(name,rollno):
        name = name
        rollno = rollno
# Doesn't work because doesn't know which object's name you mean.
```

Example with self
---

```python
class Bank:
    def __init__(self,name,acc_no):
        self.name = name
        self.acc_no = acc_no

sbi = Bank("XYZ",1234)
print(sbi.name)
print(sbi.acc_no)
    
```
